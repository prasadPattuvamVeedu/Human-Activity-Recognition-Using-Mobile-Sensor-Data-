<h1 style="color:#1565C0;">2 Outlier Detection</h1>

<h3 style="color:#E65100;">2.1 IQR Method</h3>

```python
df_clean = df.copy()
df_clean.head()
```

```python
columns.remove('Activity')
```

```python
Q1 = df_clean[columns].quantile(0.25)
Q3 = df_clean[columns].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("Lower:", lower)
print("Upper:", upper)


mask = ((df_clean[columns] >= lower) &
        (df_clean[columns] <= upper)).all(axis=1)

df_clean = df_clean[mask]
```

```python
df_clean.shape[0]
```

```python
for col in columns:
        plt.figure(figsize=(4,4))
        plt.boxplot(df_clean[col])
        plt.xlabel(col)
        plt.ylabel(col)
        plt.show()
```

```python
df_clean.shape[0]
```

<h1 style="color:#1565C0;">3 hypothesis testing</h1>

<h3 style="color:#E65100;">3.1 ANNOVA Test</h3>

```python
from scipy.stats import f_oneway
```

```python
column_annova=list(df_clean.columns)
```

```python
column_annova.remove('timestamp')
column_annova.remove('Activity')
column_annova
```

```python
df_clean['Activity']
```

```python

for col in column_annova:
        groups=groups = [
        df_clean[df_clean['Activity'] == activity][col]
        for activity in df_clean['Activity'].unique()
    ]
        f_stat, p_value = f_oneway(*groups)
        
        print(col)
        print("F-statistic:", f_stat)
        print("P-value:", p_value)
        print()
               
```

<h3 style="color:#2E7D32;">Inference</h3>

<p>
One-way ANOVA was performed to examine whether the mean values of the sensor features differed significantly across activity groups. The results indicate that accX, accY, and accZ have statistically significant differences across activities (p < 0.05), with accX and accY showing particularly strong evidence of differences. In contrast, gyroX, gyroY, and gyroZ do not show statistically significant differences across the activity groups (p > 0.05). Therefore, the accelerometer features appear to provide stronger discriminatory information for distinguishing between the activities than the gyroscope features in this analysis..
</p>

<h3 style="color:#E65100;">3.2 Mann_WhitneyTest</h3>

```python
from scipy.stats import mannwhitneyu
```

```python
columns_whitney = column_annova
```

```python
mann_df = pd.DataFrame()
for i,col in enumerate(columns_whitney):
    groups = [df_clean.loc[df_clean["Activity"] == uni][col] for uni in df_clean["Activity"].unique()]
    stat, p = mannwhitneyu(*groups)
    mann_df.loc[i,"column"] = col
    mann_df.loc[i,"stat_value"] = stat
    mann_df.loc[i,"p_value"] = p
    if p <= 0.05:
        mann_df.loc[i,"significant"] = "significant"
    else:
        mann_df.loc[i,"significant"] = "Not significant"
```

```python
mann_df 
```

<h3 style="color:#2E7D32;">Inference</h3>

<p>
The Mann–Whitney U test was performed to determine whether the distributions of the sensor features differed significantly between the two activity groups. The results show that accX, accY, and accZ are statistically significant (p < 0.05), indicating that their distributions differ between the activity groups. In contrast, gyroX, gyroY, and gyroZ are not statistically significant (p > 0.05), suggesting no significant difference in their distributions between the groups. Overall, the accelerometer features show stronger evidence of distinguishing the two activity classes than the gyroscope features.
</p>

<h1 style="color:#1565C0;">10 Model Selection and Training</h1>

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB

# for unscaled
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
model_detail = [
    {
        "name": "Logistic Regression",
        "model": LogisticRegression(
            max_iter=1000,
            random_state=42
        ),
        "parameter": [
            {
                "solver": ["liblinear"],
                "penalty": ["l1", "l2"],
                "C": [0.1, 1, 10]
            }
        ],
    },

    {
        "name": "Support Vector Classifier",
        "model": SVC(
            kernel="linear",
            C=1,
            cache_size=2000,
            random_state=42
        ),
        "parameter": {}
    },

    {
        "name": "K-Nearest Neighbors",
        "model": KNeighborsClassifier(),
        "parameter": {
            "n_neighbors": [5, 15],
            "weights": ["uniform", "distance"],
            "metric": ["euclidean", "manhattan"]
        },
    },

    {
        "name": "Gaussian Naive Bayes",
        "model": GaussianNB(),
        "parameter": {
            "var_smoothing": [1e-9]
        },
    },

    {
        "name": "Decision Tree",
        "model": DecisionTreeClassifier(
            random_state=42
        ),
        "parameter": {
            "criterion": ["gini", "entropy"],
            "max_depth": [5, 10],
            "min_samples_split": [2, 10],
            "min_samples_leaf": [1, 5]
        },
    },

    {
        "name": "Random Forest",
        "model": RandomForestClassifier(
            random_state=42,
            n_jobs=-1
        ),
        "parameter": {
            "n_estimators": [100],
            "max_depth": [10, None],
            "min_samples_split": [2],
            "max_features": ["sqrt"]
        },
    },

    {
        "name": "XGBoost",
        "model": XGBClassifier(
            random_state=42,
            eval_metric="logloss",
            n_jobs=-1
        ),
        "parameter": {
            "n_estimators": [100],
            "max_depth": [3, 5],
            "learning_rate": [0.1]
        },
    }
]
```

```python
best_estimator = {}
for model in model_detail:
    print(f"{'Model selection':<20}: {model['name']}")
    gr = GridSearchCV(estimator=model["model"],param_grid=model["parameter"], scoring="accuracy",cv=5,n_jobs=-1)
    gr.fit(x_train_resample,y_train_resample)
    best_estimator[model["name"]] = {"estimator":gr.best_estimator_}
```

```python
best_estimator
```

```python
prediction={}
for name,est in best_estimator.items():
   print(name)
   print(est)
   
   for _,model_est in est.items():
             prediction[name]=model_est.predict(x_test_scaled)
           
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, df_prediction["Random Forest"].values)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=[0, 1]
)

disp.plot()
plt.title("Confusion Matrix")
plt.show()
```
