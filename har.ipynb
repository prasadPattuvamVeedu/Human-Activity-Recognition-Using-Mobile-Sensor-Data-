<h1 style="color:#1565C0; text-align:center;">
    Human Activity Recognition (HAR)
</h1>

<p>
<strong>Human Activity Recognition (HAR)</strong> is a critical task in fields
such as health monitoring, sports analysis, and mobile computing. Using mobile
devices equipped with <strong>Inertial Measurement Unit (IMU)</strong> sensors,
such as accelerometers and gyroscopes, it is possible to detect and classify
human movements in real-time.
</p>

<p>
In this project, we use accelerometer and gyroscope data collected from mobile
devices at <strong>King Saud University</strong> to classify two basic human
activities:
</p>

<ul>
    <li><strong>Standing (Stop)</strong></li>
    <li><strong>Walking</strong></li>
</ul>


<h2 style="color:#E65100;">🎯 Goal / Objective</h2>

<p>
The main objective of this project is to develop a
<strong>machine learning model</strong> that can recognize and classify human
activities based on sensor data collected from a mobile device.
</p>

<p>The project aims to:</p>

<ul>
    <li>Analyze and preprocess the sensor data to make it suitable for machine learning.</li>
    <li>Build and evaluate machine learning models to classify activities as standing or walking.</li>
    <li>Optimize the model for accurate and robust performance with limited and noisy data.</li>
    <li>Explore the potential for real-time or near real-time activity recognition on mobile devices.</li>
</ul>


<h2 style="color:#2E7D32;">⚙️ Further Action</h2>

The model will be further optimized to ensure robust performance even when the

<h2 style="color:#00838F;">📊 Dataset Description</h2>

<p>
The dataset contains sensor measurements collected from a mobile device.
The main features include:
</p>

<ul>
    <li>
        <strong>accX, accY, accZ</strong> – Acceleration measurements along
        the X, Y, and Z axes in m/s² from the accelerometer.
    </li>
    <li>
        <strong>gyroX, gyroY, gyroZ</strong> – Angular velocity measurements
        around the X, Y, and Z axes in rad/s from the gyroscope.
    </li>
    <li>
        <strong>Timestamp</strong> – Time at which the sensor measurement
        was recorded.
    </li>
    <li>
        <strong>Activity</strong> – The target variable representing the
        activity being performed.
    </li>
</ul>

<h3 style="color:#00838F">1.1.2 Boxplot</h3>

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df.columns
```

```python
columns = list(df.columns)
for col in ['timestamp', 'Activity']:
    columns.remove(col)
columns
```

```python
fig, ax = plt.subplots(3,2, figsize=(6,6))
for i,col in enumerate(columns):
    ax.flat[i].boxplot(x=df[col])
    ax.flat[i].set_title(f"{col}")
    ax.flat[i].set_xlabel(col)
plt.tight_layout()
plt.show()
```

<h3 style="color:#2E7D32;">Inference</h3>

<p>
The box plots show the distribution and presence of potential outliers in the
accelerometer and gyroscope variables. All six sensor variables contain
several observations outside the whiskers, indicating the presence of
potential outliers. The accelerometer variables, particularly
<strong>accX</strong> and <strong>accY</strong>, show noticeable variation,
while <strong>accZ</strong> has a wider overall range. The gyroscope variables
are mostly centered around zero but also contain several extreme values.
These outliers may represent sudden movements or variations in sensor
readings and should be examined before model training.
</p>

<h3 style="color:#00838F">1.1.3 Distribution analysis</h3>

```python
fig, ax = plt.subplots(3,2, figsize=(6,6))
for i,col in enumerate(columns):
    ax.flat[i].hist(x=df[col],bins=30)
    ax.flat[i].set_title(f"{col}")
    ax.flat[i].set_xlabel(col)
plt.tight_layout()
plt.show()
```

<h3 style="color:#2E7D32;">Inference</h3>

<p>
The histograms show the distribution of the six sensor variables. 
<strong>accX</strong> and <strong>accY</strong> show a roughly bell-shaped distribution,
while <strong>accZ</strong> has a wider and less symmetric distribution.
The gyroscope variables <strong>gyroX</strong>, <strong>gyroY</strong>, and
<strong>gyroZ</strong> are mainly concentrated around zero, with some observations
extending toward both positive and negative values. Overall, the sensor
variables show different distribution patterns and variations, indicating
that each sensor feature may provide useful information for distinguishing
between standing and walking activities.
</p>
